```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error
from datetime import datetime, timedelta
import warnings
warnings.filterwarnings('ignore')

df['Market Cap'] = df['Market Cap'].str.replace(',', '').astype(float)
top_10 = df.sort_values('Market Cap', ascending=False).head(10).copy()

for index, row in top_10.iterrows():
    growth_rates = []
    if not pd.isna(row['7d %']): growth_rates.append(row['7d %'])
    if not pd.isna(row['60d %']): growth_rates.append(row['60d %']/8)  
    if not pd.isna(row['90d %']): growth_rates.append(row['90d %']/12) 
    
    if len(growth_rates) > 0:
        avg_growth = np.mean(growth_rates)
    else:
        avg_growth = 0  
    
    top_10.loc[index, 'Avg Weekly Growth'] = avg_growth

def project_price(current_price, weekly_growth, years=10):
    """
    Project price based on average weekly growth rate compounded over time
    """
    weeks = years * 52
    return current_price * (1 + weekly_growth/100) ** weeks

projections = []
for index, row in top_10.iterrows():
    current_price = row['Price']
    weekly_growth = row['Avg Weekly Growth']
    
 
    conservative_growth = weekly_growth * 0.5
    conservative_price = project_price(current_price, conservative_growth)
    

moderate_growth = weekly_growth
moderate_price = project_price(current_price, moderate_growth)
    
aggressive_growth = weekly_growth * 1.5
aggressive_price = project_price(current_price, aggressive_growth)
    
projections.append({
    'Symbol': row['Symbol'],
    'Name': row['Name'],
    'Current Price': current_price,
    'Current Market Cap': row['Market Cap'],
    'Avg Weekly Growth': weekly_growth,
    '10Y Conservative Price': conservative_price,
    '10Y Moderate Price': moderate_price,
    '10Y Aggressive Price': aggressive_price,
    '10Y Conservative Market Cap': conservative_price * (row['Market Cap'] / current_price),
    '10Y Moderate Market Cap': moderate_price * (row['Market Cap'] / current_price),
    '10Y Aggressive Market Cap': aggressive_price * (row['Market Cap'] / current_price)
    })

projections_df = pd.DataFrame(projections)

# Display results
print("Top 10 Cryptocurrencies by Market Cap - 10 Year Projections")
display(projections_df[['Symbol', 'Name', 'Current Price', 'Avg Weekly Growth', 
                       '10Y Conservative Price', '10Y Moderate Price', '10Y Aggressive Price']])

# Visualization
plt.figure(figsize=(14, 8))
symbols = projections_df['Symbol']
current_prices = projections_df['Current Price']
moderate_prices = projections_df['10Y Moderate Price']

x = range(len(symbols))
width = 0.35

plt.bar(x, current_prices, width, label='Current Price', color='blue')
plt.bar([i + width for i in x], moderate_prices, width, label='10Y Projection (Moderate)', color='orange')

plt.xlabel('Cryptocurrency')
plt.ylabel('Price (USD)')
plt.title('Current vs Projected 10-Year Prices (Moderate Scenario)')
plt.xticks([i + width/2 for i in x], symbols)
plt.yscale('log')  # Use logarithmic scale due to large price differences
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()

# Market cap comparison
plt.figure(figsize=(14, 8))
current_mcaps = projections_df['Current Market Cap'] / 1e9  # Convert to billions
moderate_mcaps = projections_df['10Y Moderate Market Cap'] / 1e9

plt.bar(x, current_mcaps, width, label='Current Market Cap', color='blue')
plt.bar([i + width for i in x], moderate_mcaps, width, label='10Y Projection (Moderate)', color='green')

plt.xlabel('Cryptocurrency')
plt.ylabel('Market Cap (USD Billions)')
plt.title('Current vs Projected 10-Year Market Caps (Moderate Scenario)')
plt.xticks([i + width/2 for i in x], symbols)
plt.yscale('log')  # Use logarithmic scale
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.show()

# Additional analysis: Polynomial regression for Bitcoin (as an example)
btc = df[df['Symbol'] == 'BTC'].iloc[0]
years = np.array([1/12, 3/12, 6/12, 1, 2]).reshape(-1, 1)  # 1m, 3m, 6m, 1y, 2y growth rates
growth_rates = np.array([btc['7d %']*4, btc['30d %'], btc['60d %'], btc['90d %']*4, btc['YTD %']]).reshape(-1, 1)

# Polynomial regression
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(years)
model = LinearRegression()
model.fit(X_poly, growth_rates)

# Predict for 10 years
future_years = np.array([1, 2, 5, 10]).reshape(-1, 1)
future_X_poly = poly.transform(future_years)
predicted_growth = model.predict(future_X_poly)

plt.figure(figsize=(10, 6))
plt.scatter(years, growth_rates, color='red', label='Historical Growth Rates')
plt.plot(future_years, predicted_growth, color='blue', label='Projected Growth Trend')
plt.xlabel('Years')
plt.ylabel('Annual Growth Rate (%)')
plt.title('Bitcoin Growth Rate Projection')
plt.legend()
plt.grid(True)
plt.show()

print("\nBitcoin Growth Rate Projections:")
for year, growth in zip(future_years.flatten(), predicted_growth.flatten()):
    print(f"{year} years: {growth:.2f}% annual growth")
    projected_price = btc['Price'] * (1 + growth/100) ** year
    print(f"Projected price: ${projected_price:,.2f}")
```
